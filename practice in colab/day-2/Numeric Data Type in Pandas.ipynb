```python
import pandas as pd
```

```python
data = {
    'Age':[25, 30, 26, 28, 20, 48, 28],
    'Salary':[54000, 60000, 45000, 50000, 61000, 49000, 54000]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df.info()
```

```python
df.dtypes
```

```python
df['Age'] = df['Age'].astype('int32')
df['Salary'] = df['Salary'].astype('int32')
```

```python
print(df)
```

```python
print(df.dtypes)
```

```python
df.info()
```

```python
df['Age'] = df['Age'].astype('int16')
df['Salary'] = df['Salary'].astype('int16')
```

```python
print(df)
print(df.info())
print(df.dtypes)
```
